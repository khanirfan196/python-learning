<h3>Full Implementation of Stack using List<h3>


```python
stack = []

# push function
def push_stack(e):
  stack.append(e)
  print("Added element: " + str(e))
  print(stack)

# peek or top function
def peek_stack():
  if len(stack) > 0:
    print("Top element: " + str(stack[-1]))
  else:
    print("Stack is empty")

# pop function
def pop_stack():
  if len(stack) > 0:
    e = stack.pop()
    print("Removed element: " + str(e))
  else:
    print("Stack is empty")

# isEmpty function
def isempty_stack():
  if len(stack) == 0:
    print("Stack is empty")
  else:
    print("Stack: " + str(stack))

# Loop to perform operations on Stack

print("1. Push element")
print("2. See last added element")
print("3. Remove element")
print("4. Check Stack elements")
print("5. Exit")

while True:
  choice = int(input("Enter Choice: "))
  if choice == 1:
    e = input("Enter element: ")
    # validation can be added for element (must be int)
    push_stack(e)
  elif choice == 2:
    peek_stack()
  elif choice == 3:
    pop_stack()
  elif choice == 4:
    isempty_stack()
  elif choice == 5:
    print("Exited Program")
    break
  else:
    print("Wrong input. Program terminated")
    break
```
