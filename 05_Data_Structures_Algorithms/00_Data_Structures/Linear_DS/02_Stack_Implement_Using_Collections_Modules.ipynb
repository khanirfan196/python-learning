```python

# Stack can be implemented using python's collections module.
    # collections has deque class which is queue where elements can be added from both the sides.
    # Using deque we can implement stack.
# Another way to implement Stack is by using python's Queue module.
    # this module has class LifoQueue.
    # here methods are different.
      # Ex: put() is use to push() elements.
      # get() to remove last element.
    # When stack gets empty and get() method is called, this method doesn't print or give any message.
    # in this case, we have to use timeout parameter and set it to 1.
    # ** Test this scenario below.
```

<h3>Implementation of Stack using Dequeu and LifoQueu classes</h3>

```python

import collections as cs
```

```python
stack = cs.deque()

print(stack)
```

```python
stack.append(10)
stack.append(20)
stack.append(30)
```

```python
print(stack)
```

```python
stack.pop()
```

```python
print(stack)
```

```python
not stack
```

```python
import queue as qu
```

```python
stack = qu.LifoQueue()

# While calling LifoQueue() class we can pass int parameter to set the lenght of the object (stack).
# stack = qu.LifoQueue(10) ---> only 10 elements can be inserted in the object (stack).
```

```python
print(stack)
```

```python
stack.put(10)
stack.put(20)
stack.put(30)

# put acts as append()

print(stack)
```

```python
# print elements of the object (stack)

print(stack.queue)
```

```python
stack.get()

# get() acts as pop() - removes last element
```

```python
stack.get()
```

```python
stack.get()
```

```python
# the below method stops execution when the list is empty.

stack.get()
```

```python
# to get response we can pass timeout parameter with value in seconds.

stack.get(timeout=1) # 1 second
```
