```python

# Priority Queue is same as Queue but elements will have priority based on their own value or separte value.
# Ex: q = [10, 20, 30] --> here 10, 20, 30 are prioritized based on their values itself. Meaning lower the value higher the priority.
# Ex: q = [30, 20, 10] --> here 30, 20, 10 are prioritized based on their values itself, meaning higher the value higher the priority.
# Ex: q = [(1, 10), (2, 20), (3, 30)] --> here the values hs their priority, 10 has priority value of 1. So based on their priority value
# operations can be performed. If 1, 2, 3 are the order of priority, their associated elements are removed 10, 20, and then 30.
# Implementing priority queue with list is simple. Same methods but list should be sorted the way we want to prioritize elements.
# Its better to implement priority queue using PriorityQueue() class from queue module.
```

```python

import queue as qu
```

```python
pqueue = qu.PriorityQueue()

print(pqueue)
```

```python

pqueue.put(10)
pqueue.put(30)
pqueue.put(70)
pqueue.put(50)
pqueue.put(20)

print(pqueue.queue)
```

```python

pqueue.get()

print(pqueue.queue)

# Here the pqueue got sorted out in ascending order and element 10 is removed.
# By default PriorityQueue() class orders elements in ascending order.
# To order in descending, heapq module should be implement in a user defined Class.
# Or simplest way to implement is by having -1 as priority for all elements.
```

```python

# Reverse order sort for priority Queue

pqr = qu.PriorityQueue()
```

```python

pqr.put((-1, 10))
pqr.put((-2, 20))
pqr.put((-3, 30))

print(pqr.queue)

```

```python
# now removing elements

pqr.get()

print(pqr.queue)

# So element 30 got removed. Though it is highest, but negative -3 priority value makes it smaller.
# also the queue got sorted out based on negative priority numbers.
```
