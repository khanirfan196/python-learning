```python

# Queue implementation using Dequeue class from collections module.
# In Dequeu, elements can be added and removed at both ends.
# append() and pop() methods are same as list. But Dequeue provides
# appendleft() and popleft() to manipulate elements at other end.
```

```python

import collections as cs
```

```python

queue = cs.deque()

print(queue)
```

```python
# Write code to fully implement Queue using deque() class.

queue.append(10)
queue.append(20)
queue.append(30)
```

```python
print(queue)
```

```python
queue.popleft()
```

```python
print(queue)
```

```python
# While implementing queue concept using dequeu() class following methos should be use:
  # approach 1:
      # append() with popleft()
      # appendleft() with pop()
  # Each set of methods adds elements at one side and removes at other.
```

```python

# Queue can also be implemented using queue() class from Queue module
# put() and get() methods are used. (append() and pop())
# There are parameters provided for the above function by this class.
# Ex: put(timeout), get(timeout) etc.
# queue() class is a straight forward implementation for queues.
```

```python

import queue as qu
```

```python
queue = qu.Queue()

print(queue)
```

```python

queue.put(10)
queue.put(20)
queue.put(30)
```

```python
print(queue.queue)
```

```python
queue.get()
```

```python
print(queue.queue)
```

```python
queue.get(block=False, timeout=1)

# timeout = 1 is 1 second
# block = False is set only if we want queue show error when no element is available.
# if set to True, it will wait until an element is available but the combination of parameters should be True and 1.
```
