```python

# A Queue is similar to a list or real time queue where people stand in line.
# It is based on FIFO or LILO -- First In First Our or Last In Last Out.
# Here the elements are added in one end and removed at other.
# The elements added to Queue at one end is called back/rear end and the other is called as front/head.
# Adding elements is called Enqueue and removing Dequeue

# Methods
# enqueue() -- append()
# dequeue() -- pop()
# isFull() -- checks whether queue is full -- only when Queue limit is set.
# isEmpty() -- not list or len(lst) == 0
```

<h3>Queue implementation using List</h3>

```python
queue = []

# Enqueue
# Dequeue
# Show Queue

# enqueue
def enqueue(e):
  queue.append(e)
  print("Added element: " + str(e))

def dequeue():
  if len(queue) > 0:
    e = queue.pop(0)
    print("Removed element: " + str(e))
  else:
    print("Queue is empty")

def displayQueue():
  if len(queue) > 0:
    print(queue)
  else:
    print("Queue is empty")

print("1. Add element to Queue")
print("2. Remove element in Queue")
print("3. Show Queue")
print("4. Exit")

while True:
  choice = int(input("Enter Operation number: "))
  if choice == 1:
    e = int(input("Enter element to add: "))
    enqueue(e)
  elif choice == 2:
    dequeue()
  elif choice == 3:
    displayQueue()
  elif choice == 4:
    print("Exited Application")
    break
  else:
    print("Wrong Input. Terminated!")
    break


```

<h3>Same implementation but by using insert() and pop methods</h3>

```python
queue = []

# Enqueue
# Dequeue
# Show Queue

# enqueue
def enqueue(e):
  queue.insert(0, e) # insert() adds element at index position 0.
  print("Added element: " + str(e))

def dequeue():
  if len(queue) > 0: # or if not queue
    e = queue.pop() # always delete from other side of list
    print("Removed element: " + str(e))
  else:
    print("Queue is empty")

def displayQueue():
  if len(queue) > 0:
    print(queue)
  else:
    print("Queue is empty")

print("1. Add element to Queue")
print("2. Remove element in Queue")
print("3. Show Queue")
print("4. Exit")

while True:
  choice = int(input("Enter Operation number: "))
  if choice == 1:
    e = int(input("Enter element to add: "))
    enqueue(e)
  elif choice == 2:
    dequeue()
  elif choice == 3:
    displayQueue()
  elif choice == 4:
    print("Exited Application")
    break
  else:
    print("Wrong Input. Terminated!")
    break
```
