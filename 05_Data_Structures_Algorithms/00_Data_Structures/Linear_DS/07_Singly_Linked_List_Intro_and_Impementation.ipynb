```python

# A singly or single linked list is a single node with data and reference to next node.
# It can perform several operations. But two operations of adding, removing, and traversal could be focus.
# Adding or Deletion of element can be done at:
    # beginning of the linkedlist
    # end of the linkedlist
    # between linkedlist
# Follow the steps through tutorial to insert new nodes at above positions.

# ** In deletion, we do not technically delete nodes, just update their references to other nodes.
# While deleting first node, the head variable value should just point to second node, which makes
# the second node as the first node of the list.
# While deleting at the end, just update the reference in node just befoer the last one to null, which breaks the link to last node
# and the updated node with ref null becomes the last node.
# While deleting in between, just point the reference to the node which is after the node to be deleted. Just breaking the link!!

# Traversal of list is nothing but reading all elements in linked list.
# For traversal we should always start with the head variable. If not null, proceeding for first node, and then so on.
# Otherwise, if head is null, the list is empty.

```

<h3>Implementation of Linked List using OOPS</h3>

```python

class Node:
  def __init__(self, data):
    self.data = data
    self.ref = None

class LinkedList:

  def __init__(self):
    self.head = None

  def addnodeatbeginning(self, data):
    new_node = Node(data)
    new_node.ref = self.head
    self.head = new_node

  def addnodeatend(self, data):
    append_node = Node(data)
    n = self.head
    if n is None:
      self.head = append_node
    else:
      while n.ref is not None:
        n = n.ref
      n.ref = append_node

  def addnodebefore(self, data, X):
    if self.head is None:
      print("Node does not exists")
      before_node = Node(data)
      before_node.ref = self.head
      self.head = before_node
    s = self.head
    while s.ref is not None:
      if s.ref.data == X:
        print("Node exists")
        break
      s = s.ref

    if s.ref is None:
      print("Node does not exists")
    else:
      before_node = Node(data)
      before_node.ref = s.ref
      s.ref = before_node


  def addnodeafter(self, data, X):
    f = self.head
    while f is not None:
      if f.data == X:
        print("Node exists")
        break
      f = f.ref

    if f is None:
      print("Node does not exists")
    else:
      after_node = Node(data)
      after_node.ref = f.ref
      f.ref = after_node


  def print_LL(self):
    n = self.head
    if n is None:
      print("Linked list is empty.")
    else:
      while n is not None:
        print(str(n) + " : " + str(n.data) + "----> Ref: " + str(n.ref))
        n = n.ref


link = LinkedList()

link.addnodeatbeginning(10)
link.addnodeatend(20)
link.addnodeatend(30)
link.addnodeatend(40)
link.addnodeatend(50)

link.print_LL()

link.addnodebefore(70, 50)


link.print_LL()
```
