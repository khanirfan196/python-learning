```python
class BST:
    def __init__(self, key):
        self.key = key
        self.lchild = None 
        self.rchild = None

    def insert_bst(self, data):
        if self.key is None:
            self.key = data 
            return
        if self.key == data:
            return
        else:
            if self.key > data:
                if self.lchild == None:
                    self.lchild = BST(data)
                else:
                    return self.lchild.insert_bst(data)
            else:
                if self.rchild == None:
                    self.rchild = BST(data)
                else:
                    return self.rchild.insert_bst(data)

    
    def search_bst(self, data):
        if self.key == data:
            return 'Node is found!' 

        if self.key > data:
            if self.lchild is None:
                return 'Node is not present in tree!'
            else:
                return self.lchild.search_bst(data)
        
        else:
            if self.rchild is None:
                return 'Node is not present in tree!'
            else:
                return self.rchild.search_bst(data)


    def min_node(self):
        current = self
        while current.lchild:
            current = current.lchild 
        print(f"node with smallest key is: {current.key}")


    # recursive method 
    def min_node2(self):
        if self.lchild:
            return self.lchild.min_node2()
        return self.key

    def max_code(self):
        current = self 
        while current.rchild:
            current = current.rchild 
        print(f"node with max value is {current.key}")

    def preorder(self):
        # visit root node
        # visit left node
        # visit right node
        print(self.key, end=" ")
        if self.lchild:
            self.lchild.preorder()
        if self.rchild:
            self.rchild.preorder()
    

    def inorder(self):
        # visit left
        # visit root
        # visit right 
        if self.lchild:
            self.lchild.inorder()
        print(self.key, end=" ")
        if self.rchild:
            self.rchild.inorder()


    def postorder(self):
        # visit left
        # visit right
        # visit root
        if self.lchild:
            self.lchild.postorder() 
        if self.rchild:
            self.rchild.postorder()
        print(self.key,  end=" ")


    # def printVals(self,action=None):
    #     print(action)
    #     print(f"self {self}")
    #     print(f"self key {self.key}")
    #     print(f"self lchild {self.lchild}")
    #     print(f"self rchild {self.rchild}")

    def delete(self, data, curr):
        print("delete method call")
        if self.key is None:
            print("Either Tree is deleted or empty!")
            return 

        if data < self.key:
            print("\ndata is less than key")
            if self.lchild:
                self.lchild = self.lchild.delete(data, curr)
            else:
                print("Given node is not present in the tree")
        
        elif data > self.key:
            print("\ndata is greater than key")
            if self.rchild:
                self.rchild = self.rchild.delete(data, curr)
            else:
                print("Given node is not present in the tree")
        
        else:
            if self.lchild is None:
                temp = self.rchild 
                if data == curr:
                    self.key = temp.key 
                    self.lchild = temp.lchild 
                    self.rchild = temp.rchild 
                    temp = None
                    return
                self = None
                return temp 
            if self.rchild is None:
                temp = self.lchild 
                if data == curr:
                    self.key = temp.key 
                    self.lchild = temp.lchild 
                    self.rchild = temp.rchild 
                    temp = None 
                    return 
                self = None 
                return temp 
            node = self.rchild
            while node.lchild:
                node = node.lchild
            self.key = node.key
            

def nodes_count(node):
    if node is None:
        return 0
    return 1 + nodes_count(node.lchild) + nodes_count(node.rchild)
        

       
                
        
```

```python
# Insertion operation
# To insert in a BST following conditions needs to be check
# 1. Is tree empty?
# 2. If yes, then the inserted node is the first node.
# 3. If no, then it contains one or more than one node, position of
# new node is required to find.
# 4. Position of new node could be either in left ST or right ST.
# 5. Also duplicate values needs to be tackle.
```

```python
# Search Operation
# To search or find any element in the BST, following points need to be considered:
# If element is equals to key or root node, then just return that value
# If element is less than key node, that element needs to be search in left sub tree.
# To proceed at left sub tree, left st should be check whether it is empty or not.
# Like if element is greater than node, that element needs to be search in right sub tree.
# Again check whether right st is empty or not.

```
