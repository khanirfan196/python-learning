```python
class BST:
    def __init__(self, key):
        self.key = key
        self.lchild = None 
        self.rchild = None

    def insert_bst(self, data):
        if self.key is None:
            self.key = data 
            return
        if self.key == data:
            return
        else:
            if self.key > data:
                if self.lchild == None:
                    self.lchild = BST(data)
                else:
                    return self.lchild.insert_bst(data)
            else:
                if self.rchild == None:
                    self.rchild = BST(data)
                else:
                    return self.rchild.insert_bst(data)

    
    def search_bst(self, data):
        if self.key == data:
            return 'Node is found!' 

        if self.key > data:
            if self.lchild is None:
                return 'Node is not present in tree!'
            else:
                return self.lchild.search_bst(data)
        
        else:
            if self.rchild is None:
                return 'Node is not present in tree!'
            else:
                return self.rchild.search_bst(data)

    def preorder(self):
        # visit root node
        # visit left node
        # visit right node
        print(self.key, end=" ")
        if self.lchild:
            self.lchild.preorder()
        if self.rchild:
            self.rchild.preorder()
    

    def inorder(self):
        # visit left
        # visit root
        # visit right 
        if self.lchild:
            self.lchild.inorder()
        print(self.key, end=" ")
        if self.rchild:
            self.rchild.inorder()


    def postorder(self):
        # visit left
        # visit right
        # visit root
        if self.lchild:
            self.lchild.postorder() 
        if self.rchild:
            self.rchild.postorder()
        print(self.key,  end=" ")


    def delete(self, data):
        if self.key is None:
            print("Tree is empty!")
            return 

        if data < self.key:
            if self.lchild:
                self.lchild = self.lchild.delete(data)
            else:
                print("Given node is not present in the tree")
        
        elif data > self.key:
            if self.rchild:
                self.rchild = self.rchild.delete(data)
            else:
                print("Given node is not present in the tree")
        
        else:
            if self.lchild is None:
                temp = self.rchild 
                self = None 
                return temp 
            
            if self.rchild is None:
                temp = self.lchild
                self = None 
                return temp 
            
            node = self.rchild

            while node.lchild:
                node = node.lchild 
            self.key = node.key 
            self.rchild = self.rchild.delete(node.key)
        
        return self 


        

```

```python
# Insertion operation
# To insert in a BST following conditions needs to be check
# 1. Is tree empty?
# 2. If yes, then the inserted node is the first node.
# 3. If no, then it contains one or more than one node, position of
# new node is required to find.
# 4. Position of new node could be either in left ST or right ST.
# 5. Also duplicate values needs to be tackle.
```

```python
# Search Operation
# To search or find any element in the BST, following points need to be considered:
# If element is equals to key or root node, then just return that value
# If element is less than key node, that element needs to be search in left sub tree.
# To proceed at left sub tree, left st should be check whether it is empty or not.
# Like if element is greater than node, that element needs to be search in right sub tree.
# Again check whether right st is empty or not.

```
