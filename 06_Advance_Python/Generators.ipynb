```python
def gen_num(n):
    for i in range(1, n+1):
        yield i
    
```

```python
gen = gen_num(5)
```

```python
print(next(gen))
```

```python
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```
