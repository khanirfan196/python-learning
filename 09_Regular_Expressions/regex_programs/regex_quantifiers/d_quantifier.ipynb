```python
import re
```

```python
# \d is for only digits. 
```

```python
list_of_strings = [
    "1234",
    ".{}-+=// Hello",
    """Apple is red in color
Special .{}(),*&#""",
    "0123456789",
]
```

```python
pattern = r"\d"
```

```python
for str in list_of_strings:
    regex_op = re.findall(pattern, str) 
    print(regex_op)
```

```python
# Use [] to match exactly one digit.
```

```python
pattern = r"[0-9]"
# this works same as above
```

```python
for str in list_of_strings:
    regex_op = re.findall(pattern, str) 
    print(regex_op)
```

```python
# with + it will return digits either one or more
```

```python
pattern = r"[0-9]+"
```

```python
for str in list_of_strings:
    regex_op = re.findall(pattern, str) 
    print(regex_op)

# instead of returning invidiulat element, it returned
# the whole string of digits as we use '+'
```

```python
# with * it returns digits either 0 or more than zero
```

```python
pattern = r"[0-9]*"
```

```python
for str in list_of_strings:
    regex_op = re.findall(pattern, str) 
    print(regex_op)

# it returned digits and also 0 digits elements 
# where it could not find any digit in each string
```
