```python
my_dict = {
    "name" : "mike",
    "email": "mike@gmail.com",
    "ids" : ["passport", "drivers license"],
    "age" : 30
}
```

```python
print(my_dict)
```

```python
for key, value in my_dict.items():
    print(key, value)
```

```python
for keys in my_dict.keys():
    print(keys)
```

```python
for value in my_dict.values():
    print(value)
```

```python
# copy() 

new_dict = my_dict.copy()

print(new_dict)
```

```python
# fromkeys() - create a dictionary with tuples

keys = ('fruit', 'veggie', 'spices')
values = ('apple', 'potato', 'pepper')

d_tuple = dict.fromkeys(keys, values)

print(d_tuple)

# for every key it picks up all the values. 
```
